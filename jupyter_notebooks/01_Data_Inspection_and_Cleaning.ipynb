## Validate Values Against the Source Documentation

The Kaggle description identifies this dataset as synthetic.
Its records should not be treated as observations of real people.

Check each numeric column against the ranges stated by the author.
These are dataset specifications, not clinical recommendations.

The overall health score is calculated from the lifestyle variables
using a weighted formula, then clipped to the range 0–100.
We will check that calculation separately.

The supplied description does not specify the exact thresholds
used to assign Poor, Average and Good health status labels.

Source:
https://www.kaggle.com/datasets/miadul/holistic-health-and-lifestyle-score-dataset

```python
# Code developed with ChatGPT assistance.

documented_ranges = {
    "Physical_Activity": (0, 120),
    "Nutrition_Score": (0, 10),
    "Stress_Level": (1, 10),
    "Mindfulness": (0, 60),
    "Sleep_Hours": (3, 10),
    "Hydration": (0.5, 5.0),
    "BMI": (18, 40),
    "Alcohol": (0, 20),
    "Smoking": (0, 30),
    "Overall_Health_Score": (0, 100)
}

range_results = []

for column, (lower, upper) in documented_ranges.items():
    outside_range = (df[column] < lower) | (df[column] > upper)

    range_results.append({
        "Column": column,
        "Allowed_Min": lower,
        "Allowed_Max": upper,
        "Observed_Min": df[column].min(),
        "Observed_Max": df[column].max(),
        "Outside_Range_Count": outside_range.sum()
    })

range_checks = pd.DataFrame(range_results)
range_checks
```

### Findings: Documented Range Validation

- All 10,000 records fall within the documented ranges for each
  of the ten numeric columns.
- No out-of-range values were found.
- Zero values are permitted for Physical_Activity, Nutrition_Score,
  Mindfulness, Alcohol, Smoking and Overall_Health_Score.
  They should not automatically be treated as missing values.
- Some observed ranges are narrower than the documented ranges.
  This does not indicate an error: a dataset does not need to
  contain every permitted value.
- No values require removal or correction based on these checks.
- Passing these checks confirms compliance with the stated ranges,
  not clinical validity or real-world representativeness.
- All original values have been retained.

## Verify the Overall Health Score

Recalculate the score using the formula supplied in the Kaggle
description and clip the result to the documented range of 0–100.

Compare the recalculated values with Overall_Health_Score using
a tolerance of 0.000001 to allow for tiny decimal differences.

This is a consistency check only. The original scores will not
be overwritten. Any discrepancies will be investigated before
deciding whether a correction is justified.

```python
# Code developed with ChatGPT assistance.

calculated_score = (
    0.2 * df["Physical_Activity"]
    + 5 * df["Nutrition_Score"]
    + 3 * df["Sleep_Hours"]
    + 5 * df["Hydration"]
    + 2 * df["Mindfulness"]
    - 2 * df["Stress_Level"]
    - 1.5 * df["Alcohol"]
    - 1.5 * df["Smoking"]
    - 1.2 * (df["BMI"] - 22)
).clip(lower=0, upper=100)

score_difference = (
    df["Overall_Health_Score"] - calculated_score
).abs()

score_mismatch = score_difference > 0.000001

print("Rows checked:", len(df))
print("Scores matching within tolerance:", (~score_mismatch).sum())
print("Scores outside tolerance:", score_mismatch.sum())
print("Largest absolute difference:", score_difference.max())
```

### Findings: Health Score Verification

- All 10,000 scores match the documented formula within the
  tolerance of 0.000001.
- No scores fall outside the tolerance.
- The largest absolute difference is approximately 5.68 × 10⁻¹⁴,
  consistent with floating-point rounding.
- No score corrections are required. Original values are retained.
- Overall_Health_Score is directly calculated from the nine input
  variables and clipped to 0–100. It is not an independently
  measured health outcome.
- This dependency must be communicated to the team: relationships
  with the score reflect its construction, and predicting it does
  not demonstrate prediction of real-world health outcomes.
